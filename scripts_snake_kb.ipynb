# Imports e carregamento do Grafo:
-------------------------------------------

```python
import os
import urllib.request
from rdflib import Graph, Literal, Namespace, RDF
import owlrl

# Baixar o arquivo base_cobras.ttl diretamente do GitHub se ele não existir na máquina do Colab
URL_TTL = "https://raw.githubusercontent.com/ArthurRicartte/OntoSerpentes/main/base_cobras.ttl"

if not os.path.exists("base_cobras.ttl"):
    print("Baixando 'base_cobras.ttl' para o ambiente do Google Colab...")
    try:
        urllib.request.urlretrieve(URL_TTL, "base_cobras.ttl")
        print("Download concluído!")
    except Exception as e:
        print(f"Erro ao baixar o arquivo: {e}")

# Carregar o grafo no RDFlib
g = Graph()
g.parse("base_cobras.ttl", format="turtle")

# Definir Namespaces e Prefixos globais
COBRA = Namespace("http://exemplo.org/cobras#")

PREFIXES = """
PREFIX cobra: <http://exemplo.org/cobras#>
PREFIX rdfs:  <http://www.w3.org/2000/01/rdf-schema#>
PREFIX owl:   <http://www.w3.org/2002/07/owl#>
PREFIX xsd:   <http://www.w3.org/2001/XMLSchema#>
"""

print(f"Grafo original carregado com sucesso no Colab! Total de triplas: {len(g)}")
```

-------------------------------------------
# Primeira bateria de Consultas via g.triples()
-------------------------------------------

```python
print("=" * 65)
print("PARTE 1: CONSULTAS RDFLIB COM g.triples()")
print("=" * 65)

# Consulta 1: Sujeito Fixo (Jararaca-da-Caatinga)
print("\n# [Consulta 1 - Sujeito Fixo] Detalhes da Jararaca-da-Caatinga:")
for s, p, o in g.triples((COBRA.JararacaDaCaatinga, None, None)):
    pred = p.split("#")[-1]
    obj = o.split("#")[-1] if "#" in str(o) else str(o)
    print(f"  • {pred} -> {obj}")

# Consulta 2: Objeto Fixo (Membros da Família Elapidae)
print("\n# [Consulta 2 - Objeto Fixo] Serpentes da família Elapidae:")
for s, p, o in g.triples((None, RDF.type, COBRA.Elapidae)):
    print(f"  • {s.split('#')[-1]}")

# Consulta 3: Predicado + Objeto Fixo (Habitantes da Caatinga)
print("\n# [Consulta 3 - Predicado + Objeto] Serpentes que habitam a Caatinga:")
for s, p, o in g.triples((None, COBRA.habitaBioma, COBRA.Caatinga)):
    print(f"  • {s.split('#')[-1]}")

# Consulta 4: Predicado + Objeto Fixo (Serpentes Noturnas)
print("\n# [Consulta 4 - Literal Fixo] Serpentes com hábito noturno (true):")
for s, p, o in g.triples((None, COBRA.eNoturna, Literal(True))):
    print(f"  • {s.split('#')[-1]}")

# Consulta 5: Predicado + Objeto Fixo (Dentição Opistóglifa)
print("\n# [Consulta 5 - Predicado + Objeto] Serpentes com dentição Opistóglifa:")
for s, p, o in g.triples((None, COBRA.possuiDenticao, COBRA.Opistoglifa)):
    print(f"  • {s.split('#')[-1]}")
```

-------------------------------------------
# Segunda bateria de consultas via SPARQL
-------------------------------------------

```python
print("=" * 65)
print("CONSULTAS DE SPARQL")
print("=" * 65)

# Consulta 1: Filtrar nome popular e nome científico de cobras com hábito noturno
c1 = PREFIXES + """
SELECT ?nomePopular ?nomeCientifico
WHERE {
    ?s cobra:eNoturna ?noturna ;
       cobra:nomePopular ?nomePopular ;
       cobra:nomeCientifico ?nomeCientifico .
    FILTER(?noturna)
}
ORDER BY ?nomePopular
"""
print("\n# Consulta 1 -> Serpentes com hábito noturno:")
for row in g.query(c1):
    print(f"  • {row.nomePopular} -> ({row.nomeCientifico})")

# Consulta 2: Ranking de tamanho médio acima de 1.5 metros
c2 = PREFIXES + """
SELECT ?nomePopular ?nomeCientifico ?tamanhoMedio
WHERE {
    ?s cobra:nomePopular ?nomePopular ;
       cobra:nomeCientifico ?nomeCientifico ;
       cobra:comprimentoMedioMetros ?tamanhoMedio .
    FILTER (?tamanhoMedio >= 1.5)
}
ORDER BY DESC(?tamanhoMedio)
"""
print("\n# Consulta 2 -> Ranking de tamanho médio acima de 1.5 metros:")
for row in g.query(c2):
    tamanho = float(row.tamanhoMedio)
    print(f"  • {row.nomePopular} -> ({row.nomeCientifico}) -> {tamanho:.1f} metros")

# Consulta 3: Ranking com maior alerta de acidentes
c3 = PREFIXES + """
SELECT ?nomePopular ?nomeCientifico ?nivelRisco
WHERE {
    ?s cobra:nomePopular ?nomePopular ;
       cobra:nomeCientifico ?nomeCientifico ;
       cobra:nivelRiscoAcidente ?nivelRisco .
}
ORDER BY DESC(?nivelRisco) ?nomePopular
LIMIT 5
"""
print("\n# Consulta 3 -> Top 5 cobras com maior Nível de Risco de Acidente:")
for row in g.query(c3):
    print(f"  • {row.nomePopular} ({row.nomeCientifico}) -> Nível de Risco: {row.nivelRisco}/5")

# Consulta 4: Top 3 cobras com maior diversidade de biomas
c4 = PREFIXES + """
SELECT ?nomePopular (COUNT(?bioma) AS ?qtdBiomas)
WHERE {
    ?s cobra:nomePopular ?nomePopular ;
       cobra:habitaBioma ?bioma .
}
GROUP BY ?nomePopular
ORDER BY DESC(?qtdBiomas) ?nomePopular
LIMIT 3
"""
print("\n# Consulta 4 -> Top 3 Serpentes com maior diversidade de Biomas:")
for row in g.query(c4):
    print(f"  • {row.nomePopular} -> Habita {row.qtdBiomas} biomas")

# Consulta 5: Existe cobra peçonhenta na Amazônia? (ASK)
c5 = PREFIXES + """
ASK {
    ?s a/rdfs:subClassOf* cobra:Peconhenta ;
       cobra:habitaBioma cobra:Amazonia .
}
"""

# Consulta 6: Exibindo resultado da c5
c6 = PREFIXES + """
SELECT ?nomePopular ?nomeCientifico ?familia
WHERE {
    ?s a/rdfs:subClassOf* cobra:Peconhenta ;
       a ?familia ;
       cobra:habitaBioma cobra:Amazonia ;
       cobra:nomePopular ?nomePopular ;
       cobra:nomeCientifico ?nomeCientifico .

    FILTER(?familia != cobra:Peconhenta && ?familia != cobra:Serpente)
}
ORDER BY ?nomePopular
"""
print(f"\n# Consulta 5 -> Existe alguma serpente peçonhenta na Amazônia? {g.query(c5).askAnswer}")
print("# Consulta 6 -> Serpentes peçonhentas que habitam a Amazônia:")
for row in g.query(c6):
    familia = row.familia.split("#")[-1]
    print(f"  • {row.nomePopular} ({row.nomeCientifico}) -> Família: {familia}")

# Consulta 7: Criando um alerta para todas as cobras peçonhentas (CONSTRUCT)
c7 = PREFIXES + """
CONSTRUCT {
    ?s cobra:alerta "Atenção: Espécie Peçonhenta!" .
}
WHERE {
    ?s a/rdfs:subClassOf* cobra:Peconhenta .
}
"""
print("\n# Consulta 7 (CONSTRUCT) -> Grafo de Alertas Gerado:")
for s, p, o in g.query(c7):
    nome_cobra = s.split("#")[-1]
    print(f"  • {nome_cobra} -> {o}")

# Consulta 8: INSERT DATA: Adiciona uma nota sobre a Cascavel
u1 = PREFIXES + """
INSERT DATA {
    cobra:Cascavel cobra:observacao "Possui chocalho na cauda." .
}
"""
g.update(u1)
print("\n# Consulta 8 (UPDATE INSERT) -> Observação adicionada à Cascavel:")
for s, p, o in g.triples((COBRA.Cascavel, COBRA.observacao, None)):
    print(f"  • {s.split('#')[-1]} -> {p.split('#')[-1]}: \"{o}\"")

# Consulta 9: DELETE DATA: Remove a nota adicionada
u2 = PREFIXES + """
DELETE DATA {
    cobra:Cascavel cobra:observacao "Possui chocalho na cauda." .
}
"""
g.update(u2)
print("\n# Consulta 9 (UPDATE DELETE) -> Observação removida da Cascavel:")
triplas_observacao = list(g.triples((COBRA.Cascavel, COBRA.observacao, None)))
if not triplas_observacao:
    print("  • Tripla removida com sucesso! (Nenhum resultado encontrado)")
else:
    for s, p, o in g.triples((COBRA.Cascavel, COBRA.observacao, None)):
        print(f"  • {s.split('#')[-1]} -> {p.split('#')[-1]}: \"{o}\"")

# Consulta 10: DELETE/INSERT: Atualiza o comprimento médio da Coral Verdadeira
u3 = PREFIXES + """
DELETE { cobra:CoralVerdadeira cobra:comprimentoMedioMetros ?antigo . }
INSERT { cobra:CoralVerdadeira cobra:comprimentoMedioMetros 0.85 . }
WHERE  { cobra:CoralVerdadeira cobra:comprimentoMedioMetros ?antigo . }
"""
g.update(u3)
print("\n# Consulta 10 (UPDATE DELETE/INSERT) -> Comprimento médio da Coral Verdadeira atualizado para 0.85m:")
for s, p, o in g.triples((COBRA.CoralVerdadeira, COBRA.comprimentoMedioMetros, None)):
    print(f"  • {s.split('#')[-1]} -> {p.split('#')[-1]}: {o}m")
```

# Terceira bateria de consultas testando as restrições do OWl e as inferências via OWLRL
--------------------------------------------------------------------------------


```python
print("=" * 70)
print("TESTANDO AS 5 CONSTRUÇÕES OWL DA ONTOLOGIA E EXPANSÃO")
print("=" * 70)

# ------------------------------------------------------------------------------
# Teste 1: owl:inverseOf (eHabitadoPor)
# ------------------------------------------------------------------------------
teste1 = PREFIXES + """
SELECT ?cobra WHERE {
    cobra:Caatinga cobra:eHabitadoPor ?cobra .
}
"""

print("\n1. Teste de owl:inverseOf (Bioma -> eHabitadoPor -> Cobra):")
print(f"  • Antes da inferência OWL: {len(list(g.query(teste1)))} resultados encontrados.")

# Expande o grafo aplicando a fechadura dedutiva das regras semânticas do OWL
owlrl.DeductiveClosure(owlrl.OWLRL_Semantics).expand(g)

# Salva o resultado expandido em um arquivo separado
#g.serialize(destination="base_cobras_expandida.ttl", format="turtle")
print("  • Grafo expandido salvo com sucesso em 'base_cobras_expandida.ttl'!")

resTeste1 = list(g.query(teste1))
print(f"  • Após a inferência OWL: {len(resTeste1)} resultados encontrados!")
for row in resTeste1:
    print(f"    - Caatinga eHabitadoPor -> {row.cobra.split('#')[-1]}")

# ------------------------------------------------------------------------------
# Teste 2: owl:SymmetricProperty (eSimpatricaCom)
# ------------------------------------------------------------------------------
teste2 = PREFIXES + """
SELECT ?cobra WHERE {
    cobra:CoralVerdadeira cobra:eSimpatricaCom ?cobra .
}
"""
print("\n2. Teste de owl:SymmetricProperty (Simetria Ecológica):")
print("  • Verificando se a relação simétrica foi derivada para a Coral-Verdadeira:")
for row in g.query(teste2):
    print(f"    - CoralVerdadeira eSimpatricaCom -> {row.cobra.split('#')[-1]}")

# ------------------------------------------------------------------------------
# Teste 3: owl:SymmetricProperty (temParenteProximo)
# ------------------------------------------------------------------------------
teste3 = PREFIXES + """
SELECT ?cobra WHERE {
    cobra:CorreCampo cobra:temParenteProximo ?cobra .
}
"""
print("\n3. Teste de owl:SymmetricProperty (Simetria Evolutiva/Taxonômica):")
print("  • Verificando parentesco derivado para o Corre-Campo:")
for row in g.query(teste3):
    print(f"    - CorreCampo temParenteProximo -> {row.cobra.split('#')[-1]}")

# ------------------------------------------------------------------------------
# Teste 4: owl:FunctionalProperty (nomeCientifico)
# ------------------------------------------------------------------------------
teste4 = PREFIXES + """
SELECT ?cobra (COUNT(?nome) AS ?totalNomes) WHERE {
    ?cobra cobra:nomeCientifico ?nome .
}
GROUP BY ?cobra
HAVING (?totalNomes > 1)
"""
res_func = list(g.query(teste4))
print("\n4. Teste de owl:FunctionalProperty (Unicidade do Nome Científico):")
if not res_func:
    print("  • Nenhuma instância viola o axioma FunctionalProperty (todas têm no máximo 1 nome científico).")
else:
    print("  • Alerta: Inconsistência encontrada! Instância com mais de um nome científico.")

# ------------------------------------------------------------------------------
# Teste 5: owl:disjointWith (Peconhenta vs NaoPeconhenta)
# ------------------------------------------------------------------------------
teste5 = PREFIXES + """
SELECT ?cobra WHERE {
    ?cobra a cobra:Peconhenta ;
           a cobra:NaoPeconhenta .
}
"""
res_disjoint = list(g.query(teste5))
print("\n5. Teste de owl:disjointWith (Disjunção de Classes):")
if not res_disjoint:
    print("  • Nenhuma instância viola a disjunção (nenhuma cobra é Peçonhenta e Não Peçonhenta simultaneamente).")
else:
    print("  • Alerta: Inconsistência de disjunção detectada!")
```
